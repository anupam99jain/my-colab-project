```python
import requests
from bs4 import BeautifulSoup
import tldextract
import re
from urllib.parse import urlparse

print("Packages ready!")
```

```python
def analyze_url_basic(url):

    parsed = urlparse(url)

    if not parsed.scheme:
        url = "https://" + url
        parsed = urlparse(url)

    domain = parsed.netloc
    path = parsed.path
    query = parsed.query

    features = {}

    features["URLLength"] = len(url)
    features["DomainLength"] = len(domain)

    # IP address check
    ip_pattern = r"^(?:\d{1,3}\.){3}\d{1,3}$"
    features["IsDomainIP"] = int(bool(re.match(ip_pattern, domain)))

    # HTTPS
    features["IsHTTPS"] = int(parsed.scheme.lower() == "https")

    # Subdomains
    extracted = tldextract.extract(domain)
    subdomain = extracted.subdomain

    if subdomain:
        features["NoOfSubDomain"] = len(subdomain.split("."))
    else:
        features["NoOfSubDomain"] = 0

    # URL characters
    features["NoOfLettersInURL"] = sum(c.isalpha() for c in url)
    features["NoOfDegitsInURL"] = sum(c.isdigit() for c in url)

    features["NoOfEqualsInURL"] = url.count("=")
    features["NoOfQMarkInURL"] = url.count("?")
    features["NoOfAmpersandInURL"] = url.count("&")

    special_chars = set("!@#$%^&*()_+-=[]{}|;:',./<>?`~")
    features["NoOfOtherSpecialCharsInURL"] = sum(
        c in special_chars for c in url
    )

    # Ratios
    url_len = max(len(url), 1)

    features["LetterRatioInURL"] = (
        features["NoOfLettersInURL"] / url_len
    )

    features["DegitRatioInURL"] = (
        features["NoOfDegitsInURL"] / url_len
    )

    features["SpacialCharRatioInURL"] = (
        features["NoOfOtherSpecialCharsInURL"] / url_len
    )

    # Obfuscation
    obfuscated_chars = sum(
        url.lower().count(x)
        for x in ["%", "\\x", "\\u"]
    )

    features["NoOfObfuscatedChar"] = obfuscated_chars
    features["HasObfuscation"] = int(obfuscated_chars > 0)
    features["ObfuscationRatio"] = obfuscated_chars / url_len

    return features
```

```python
def analyze_webpage(url):

    if not url.startswith(("http://", "https://")):
        url = "https://" + url

    try:
        response = requests.get(
            url,
            timeout=10,
            headers={
                "User-Agent": "Mozilla/5.0"
            }
        )

        html = response.text
        soup = BeautifulSoup(html, "html.parser")

        result = {}

        # Forms
        forms = soup.find_all("form")
        result["HasExternalFormSubmit"] = 0

        for form in forms:
            action = form.get("action", "")

            if action:
                parsed_action = urlparse(action)
                parsed_url = urlparse(url)

                if (
                    parsed_action.netloc
                    and parsed_action.netloc != parsed_url.netloc
                ):
                    result["HasExternalFormSubmit"] = 1

        # Submit buttons
        submit_buttons = soup.find_all(
            ["input", "button"]
        )

        result["HasSubmitButton"] = int(
            any(
                tag.get("type", "").lower() == "submit"
                or "submit" in tag.get_text().lower()
                for tag in submit_buttons
            )
        )

        # Hidden fields
        hidden_fields = soup.find_all(
            "input",
            {"type": "hidden"}
        )

        result["HasHiddenFields"] = int(
            len(hidden_fields) > 0
        )

        # Password field
        password_fields = soup.find_all(
            "input",
            {"type": "password"}
        )

        result["HasPasswordField"] = int(
            len(password_fields) > 0
        )

        # iframe
        result["NoOfiFrame"] = len(
            soup.find_all("iframe")
        )

        # Popups - basic JS indicators
        popup_keywords = [
            "alert(",
            "window.open",
            "popup"
        ]

        result["NoOfPopup"] = sum(
            html.lower().count(keyword)
            for keyword in popup_keywords
        )

        # Redirects
        result["NoOfURLRedirect"] = len(
            response.history
        )

        result["NoOfSelfRedirect"] = 0

        return result

    except Exception as e:

        print("Website analysis failed:", e)

        return {
            "HasExternalFormSubmit": 0,
            "HasSubmitButton": 0,
            "HasHiddenFields": 0,
            "HasPasswordField": 0,
            "NoOfiFrame": 0,
            "NoOfPopup": 0,
            "NoOfURLRedirect": 0,
            "NoOfSelfRedirect": 0
        }
```

```python
def extract_features(url):

    url_features = analyze_url_basic(url)
    web_features = analyze_webpage(url)

    all_features = {}

    all_features.update(url_features)
    all_features.update(web_features)

    return all_features
```

```python
import re
import ipaddress
from urllib.parse import urlparse

def extract_url_features(url):
    # Add scheme if user enters only domain
    if not re.match(r"^https?://", url, re.IGNORECASE):
        url = "https://" + url

    parsed = urlparse(url)
    domain = parsed.netloc.split("@")[-1].split(":")[0]

    # Basic counts
    url_length = len(url)
    domain_length = len(domain)

    try:
        ipaddress.ip_address(domain)
        is_domain_ip = 1
    except ValueError:
        is_domain_ip = 0

    # Subdomains
    domain_parts = domain.split(".")
    no_of_subdomain = max(len(domain_parts) - 2, 0)

    # Character counts
    letters = sum(c.isalpha() for c in url)
    digits = sum(c.isdigit() for c in url)

    no_of_equals = url.count("=")
    no_of_qmark = url.count("?")
    no_of_ampersand = url.count("&")

    # Other special characters
    special_chars = sum(
        not c.isalnum() and c not in [".", "/", "=", "?", "&", "-", "_"]
        for c in url
    )

    # Obfuscation indicators
    obfuscated_chars = url.count("%")
    has_obfuscation = int(obfuscated_chars > 0)

    # Ratios
    letter_ratio = letters / url_length if url_length else 0
    digit_ratio = digits / url_length if url_length else 0
    special_ratio = special_chars / url_length if url_length else 0
    obfuscation_ratio = obfuscated_chars / url_length if url_length else 0

    # HTTPS
    is_https = int(parsed.scheme.lower() == "https")

    features = {
        "URLLength": url_length,
        "DomainLength": domain_length,
        "IsDomainIP": is_domain_ip,
        "NoOfSubDomain": no_of_subdomain,
        "HasObfuscation": has_obfuscation,
        "NoOfObfuscatedChar": obfuscated_chars,
        "ObfuscationRatio": obfuscation_ratio,
        "NoOfLettersInURL": letters,
        "LetterRatioInURL": letter_ratio,
        "NoOfDegitsInURL": digits,
        "DegitRatioInURL": digit_ratio,
        "NoOfEqualsInURL": no_of_equals,
        "NoOfQMarkInURL": no_of_qmark,
        "NoOfAmpersandInURL": no_of_ampersand,
        "NoOfOtherSpecialCharsInURL": special_chars,
        "SpacialCharRatioInURL": special_ratio,
        "IsHTTPS": is_https
    }

    return features


print("URL feature extractor created successfully!")
```

```python
def get_risk_level(phishing_probability):
    if phishing_probability >= 0.70:
        return "HIGH RISK"
    elif phishing_probability >= 0.40:
        return "MEDIUM RISK"
    else:
        return "LOW RISK"


def explain_prediction(feature_vector, model):
    importances = model.feature_importances_

    values = feature_vector.iloc[0]

    feature_importance = pd.DataFrame({
        "Feature": feature_vector.columns,
        "Value": values.values,
        "Importance": importances
    })

    # Top 5 model-important features
    top_features = feature_importance.sort_values(
        "Importance",
        ascending=False
    ).head(5)

    return top_features


print("Risk analysis functions created successfully!")
```

```python
def analyze_url_from_features(feature_vector):
    # Prediction
    prediction = saved_model.predict(feature_vector)[0]
    probabilities = saved_model.predict_proba(feature_vector)[0]

    phishing_probability = probabilities[0]
    legitimate_probability = probabilities[1]

    # Risk level
    risk_level = get_risk_level(phishing_probability)

    # Important features
    top_features = explain_prediction(
        feature_vector,
        saved_model
    )

    # Result
    result = {
        "prediction": (
            "LEGITIMATE"
            if prediction == 1
            else "PHISHING"
        ),
        "phishing_probability": round(
            phishing_probability * 100, 2
        ),
        "legitimate_probability": round(
            legitimate_probability * 100, 2
        ),
        "risk_level": risk_level,
        "top_features": top_features[
            "Feature"
        ].tolist()
    }

    return result


print("ShopSafe AI analysis function created successfully!")
```
